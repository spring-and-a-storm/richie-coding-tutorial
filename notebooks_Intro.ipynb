## Your First Program

The first thing you usually learn to do in a programming course is print the phrase "Hello, world!" In Python, this is very intuitive.

In Jupyter notebooks, you can run the code by clicking on the code cell, holding Shift, and pressing Enter.

```python
print("Hello, world!")
```

Feel free to change the phrase inside the quotation marks to anything you'd like.

print( ) is a method. A method is a block of code that (usually) takes an input, and returns an output accordingly. Python has a bunch of built-in methods like this.

The input is what goes inside of the parentheses. In this case, the input is "Hello, world!" Try doing this without the quotation marks.

```python
print(Hello, world!)
```

This doesn't work. The computer sees the input (Hello, world!) and doesn't recognize it as text. It thinks Hello and world are both supposed to be variables, and the comma separating them tells it that these variables are supposed to satisfy two different parameters of the method. (We'll get into what this means later.) The exclamation point is nonsensical here, so it returns a **syntax error**. You are going to encounter a lot of syntax errors in your time as a programmer.

## Syntax

Learning Syntax could be equated to learning the grammar system of a foreign language, assuming that grammar system was required for you to be understood at all. *(Which would be silly. It would be like if missing an Oxford comma caused people to think you were spouting complete nonsense.)* Syntax, as you'll come to learn, is the format you have to put your code in for the computer to properly understand it.

To print a simple phrase like this, it needs to be written as a **string**, which is a group of characters that you can't do math with. You define a string with "quotation marks".

Imagine that when you call the print( ) method *("calling" the method is what it's... called... when you tell your computer to run the method,)* your computer defines an invisible variable as whatever you put inside the parentheses as your input. It then tries to print the contents of that variable. Variables can be defined as several different kinds of things, including strings, numbers, lists, and other variables -- or, more specifically, the contents of other variables. The syntax for a variable is plain text with no spaces. When the method tries to define its invisible variable as Hello, that doesn't work, because Hello is a variable with nothing in it.

In this case, that actually doesn't matter, because what the computer noticed first was that exclamation point, It doesn't even know what that's supposed to be.

## Data Types

The main four data types are **strings, integers, floats, and bools.** These four things all serve different purposes. A variable can be defined as any of them.

We already talked about strings. You can edit their text, also.

An integer is a whole number. You can do math with integers.

Floats, or floating-point values, can be any real number, including integers and decimal numbers. You can do math with floats, but it may be best to avoid it if possible. Computers do not like floats.

A bool, or a Boolean value, is either true or false. You can use these to do logic. Boolean values are named after one George Boole, the dude who came up with the idea that you can do logic by evaluating whether statements are true or false. This was mindblowing in the 1800s.

```python
# By the way, putting a pound sign before something indicates that it's a comment.

string = "string"
integer = 1
float = 1.5
bool = True

print(string, integer, float, bool)
```

Two more things:

When writing code, try to keep it simple and readable.

When learning Object-Oriented Programming, your teacher will make you pretend that your entire program needs to be a method. It's good to get in the habit for when we actually start using methods. This is how you do that:

```python
def main():
    message = "My name is Richie!"
    print(message)

main()
```

Think about it like this: When you start coding on Scratch, it makes the most sense to put together simple programs in one single group (column?) of blocks. Eventually, you're going to need to divide the code across separate columns, and even separate sprites, to write more complex programs easily. *(You also might want to divide it so it isn't painful to look at.)* That's what Object-Oriented Programming is all about.
